```python
# File operations
# handle=open(filename, mode)
# rstrip() removes the extra \n character at the end of each line
name='Hello\nWorld'
print(name)
count=0
fhand=open('file.txt')
#print(fhand)
for line in fhand:
    line=line.rstrip()
    print(line)
    count=count+1
print('Line Count:',count)
```

```python
# Reading the whole file
xfile=open('file.txt')
inp=xfile.read()
print(len(inp))
print(inp)
```

```python
# Searching through a file
print('The \'if\' logic')
fhand=open('file.txt')
for line in fhand:
    line=line.rstrip()
    if line.startswith('H'):
        print(line)
        
# same logic
print()
print('The \'if not\' logic')
fhand=open('file.txt')
for line in fhand:
    line=line.rstrip()
    if not line.startswith('H'):
        continue
    print(line)
    
# using the 'in' operator
print()
print('The \'in operator\' logic')
fhand=open('file.txt')
for line in fhand:
    line=line.rstrip()
    if not 'H' in line:
        continue
    print(line)
```

```python
# prompt for filename
fname=input('Enter the filename: ')
try:
    fhand=open(fname)
except:
    print('File cannot be opened',fname)
    quit()
    
count=0
for line in fhand:
    if line.startswith('H'):
        count=count+1
print('There are',count,'lines beginning with \'H\' in',fname)
```
