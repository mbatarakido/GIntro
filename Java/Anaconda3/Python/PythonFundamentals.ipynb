```python
# try - except concept, works like try - catch in Java
astr = 'Hello Bob 2'
try:
    istr = int(astr)
except:
    istr = -1
print('First ', istr)

bstr = '123'
try:
    cstr = int(bstr)
except:
    cstr = -1
print('Second ', cstr)
```

```python
# find smallest number in a loop
smallest = None
print('Before')
values = [9,5,2,34,7,87,405,1, 12]
for value in values:
    if smallest is None:
        smallest = value
    elif value < smallest:
        smallest = value
    print(smallest, value)
print('After', smallest)
```
